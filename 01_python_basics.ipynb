```python
def myfunc(func):
    def wrapper():
        return func().upper()
    return wrapper

@myfunc
def greet():
    return "hello"

print(greet())


```

Multiple Decorator Calls

```python
def changecase(func):
    def wrapper():
        return func().upper()
    return wrapper

@changecase
def greet1():
    return "Hello"

@changecase
def greet2():
    st = "how are you ? "
    return st

print(greet1())
print(greet2())
```
