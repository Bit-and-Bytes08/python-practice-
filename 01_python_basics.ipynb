```python
def myfunc(func):
    def wrapper():
        return func().upper()
    return wrapper

@myfunc
def greet():
    return "hello"

print(greet())


```

Multiple Decorator Calls

```python
def changecase(func):
    def wrapper():
        return func().upper()
    return wrapper

@changecase
def greet1():
    return "Hello"

@changecase
def greet2():
    st = "how are you ? "
    return st

print(greet1())
print(greet2())
```

Arguments in Decorators

```python
def changecase(x):
    def wrapper(x):
        return x.upper()
    return wrapper

@changecase
def inner(x):
    return x

print(inner("hello"))
```

```python
def myfunc(fname):
    def inner_func(func):  # Renamed for clarity to avoid shadowing
        print(fname)
        print(func())
    return inner_func      # <-- Added this critical return statement

def inner():
    return "usman"

myfunc("ali")(inner)
# Output:
# ali
# usman
```

```python
def dec1(func):
    def inner(*args, **kargs):
        print("before")
        func(*args, **kargs)
        print("after")
    return inner

def dec2(func):
    def inner(*args, **kargs):    
        print("=========")
        func(*args, **kargs)
        print("=========")
    return inner

def dec3(func):
    def inner(*args, **kargs): 
        print("Welcome")
        func(*args, **kargs)
        print("Good bye")
    return inner


a = 2
b = 3

@dec3
@dec2
@dec1
def add(a, b):
    print(f"add: {a+b}")

add(a,b)
```

# recurrsion function

```python
def countdown(n):
    if n <= 0:
        print("done")
    else:
        print(n)
        countdown(n-1)
countdown(5)
```

```python
def factorial(n):
    if n == 0:
        print("Factorial done")
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))
```

```python
# fabonaci series

def fabonacii(lis, length, i):
    if i>=length:
        return lis
    else:
        lis.append(lis[len(lis) - 1] + lis[len(lis) - 2])
        i = i + 1
        return fabonacii(lis, length, i)

lis = [0, 1]
print(fabonacii(lis, 9, 2))
print(len(lis))
```

```python
# sum of all elements 
def sum(n):
    if n == 0:
        return 0
    else:
        return n + sum(n-1)
print(sum(90))
```

```python
numbers = [1,2,3,4,5]
def sum(n):
    if len(n) <= 0:
        return 0
    else:
        return n[0] + sum(n[1:])
print(sum(numbers))
```

# Generators

```python
# simple generator function
def my_generator():
    yield 1
    yield 2
    yield 3

it = my_generator()
print(it)
print(it.__next__())
print(next(it))
print(next(it))
```

```python
# list comprehension
li = [n for n in range(10)]
print(li)

# generator expression
gen_exp = (n for n in range(10))
print(gen_exp)
```
