```python
def myfunc(func):
    def wrapper():
        return func().upper()
    return wrapper

@myfunc
def greet():
    return "hello"

print(greet())


```

Multiple Decorator Calls

```python
def changecase(func):
    def wrapper():
        return func().upper()
    return wrapper

@changecase
def greet1():
    return "Hello"

@changecase
def greet2():
    st = "how are you ? "
    return st

print(greet1())
print(greet2())
```

Arguments in Decorators

```python
def changecase(x):
    def wrapper(x):
        return x.upper()
    return wrapper

@changecase
def inner(x):
    return x

print(inner("hello"))
```

```python
def myfunc(fname):
    def inner_func(func):  # Renamed for clarity to avoid shadowing
        print(fname)
        print(func())
    return inner_func      # <-- Added this critical return statement

def inner():
    return "usman"

myfunc("ali")(inner)
# Output:
# ali
# usman
```

```python
def dec1(func):
    def inner(*args, **kargs):
        print("before")
        func(*args, **kargs)
        print("after")
    return inner

def dec2(func):
    def inner(*args, **kargs):    
        print("=========")
        func(*args, **kargs)
        print("=========")
    return inner

def dec3(func):
    def inner(*args, **kargs): 
        print("Welcome")
        func(*args, **kargs)
        print("Good bye")
    return inner


a = 2
b = 3

@dec3
@dec2
@dec1
def add(a, b):
    print(f"add: {a+b}")

add(a,b)
```
